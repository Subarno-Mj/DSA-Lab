# **Single Linked List** ---

**A singly linked list is composed of Nodes, where each node stores a data value and a reference (next) to the subsequent node in the sequence.**

```python

print("Singly Linked List---")


# Single Node structure
class Node:
    def __init__(self,data=None):
        self.data = data
        self.next = None


class Singly_Linked_List:

    def __init__(self):
        self.head = None


    def insert_at_head(self,data):
        new_node =Node(data)
        new_node.next=self.head
        self.head=new_node

    def insert_at_tail(self,data):
        new_node = Node(data)
        if not self.head:     # if list is empty
            self.head=new_node
            return

        current =self.head
        while current.next:
            current = current.next # loop terminates at last element
        current.next=new_node

    def delete(self,target):

        current =self.head

        # case 1- targer is head
        if current and current.data==target:
            self.head=current.next
            return True

        # case 2 - in middle somewhere or end
        prev = None
        while current and current.data !=target:
            prev = current
            current = current.next

        # when Target not found
        if not current:   # current becomes Null i.e. current did nt match target so target not present
           return False

        # when targer is found
        prev.next=current.next
        return True




    def search(self , target):
        current = self.head
        while current:
            if current.data==target:
                return True
            current = current.next
        return False

    def print_List(self):
        current = self.head
        while current:
            print(current.data,end="->")
            current=current.next
        print(None)


# Initialize List by creating object
sl_object = Singly_Linked_List()

# insert element
sl_object.insert_at_head(10)
sl_object.insert_at_head(20)
sl_object.insert_at_head(30)
sl_object.insert_at_head(40)
sl_object.insert_at_head(50)
sl_object.insert_at_head(100)

# print
sl_object.print_List()

x=int(input("Find an element ?"))
if sl_object.search(x):
  print("Search Successful ...")
else:
  print("Sorry Not found...!")


```

# **Doubly Linked List**

```python
# single Node structure
class Node:
    def __init__(self, data):
        self.data = data
        self.prev = None
        self.next = None


class DoublyLinkedList:
    def __init__(self):
        self.head = None

    # Insert at the beginning
    def insert_at_beginning(self, data):
        new_node = Node(data)

        if self.head is not None:
            new_node.next = self.head
            self.head.prev = new_node

        self.head = new_node

    # Insert at the end
    def insert_at_end(self, data):
        new_node = Node(data)

        if self.head is None:
            self.head = new_node
            return

        current = self.head
        while current.next:
            current = current.next

        current.next = new_node
        new_node.prev = current

    # Delete a node
    def delete(self, key):
        if self.head is None:
            return

        current = self.head

        while current and current.data != key:
            current = current.next

        if current is None:
            print("Value not found")
            return

        # If deleting the head
        if current == self.head:
            self.head = current.next
            if self.head:
                self.head.prev = None
            return

        # Update links
        if current.prev:
            current.prev.next = current.next

        if current.next:
            current.next.prev = current.prev

    # Search for a value
    def search(self, key):
        current = self.head
        while current:
            if current.data == key:
                return True
            current = current.next
        return False

    # Traverse forward
    def display_forward(self):
        current = self.head
        while current:
            print(current.data, end=" <-> ")
            current = current.next
        print("None")

    # Traverse backward
    def display_backward(self):
        if self.head is None:
            print("None")
            return

        current = self.head
        while current.next:
            current = current.next

        while current:
            print(current.data, end=" <-> ")
            current = current.prev
        print("None")


# Example usage
dll = DoublyLinkedList()

dll.insert_at_beginning(20)
dll.insert_at_beginning(10)
dll.insert_at_end(30)
dll.insert_at_end(40)

print("Forward:")
dll.display_forward()

print("Backward:")
dll.display_backward()

print("Search 30:", dll.search(30))
print("Search 50:", dll.search(50))

dll.delete(30)

print("After deleting 30:")
dll.display_forward()
```
