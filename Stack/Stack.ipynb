# **Stack Implementation**

```python
class Stack:
    print("Stack Implementation---")

    def __init__(self):
        self.items = []

    def is_empty(self):
        if len(self.items) == 0:
            return True
        else:
            return False

    def push(self, data):
        self.items.append(data)

    def pop(self):
        if self.is_empty():
            print("Stack is empty...")
            return
        else:
            return self.items.pop()

    def peek(self):
        if self.is_empty():
            print("Stack is empty...")
            return
        else:
            return self.items[-1]

    def print_stack(self):
        for i in self.items:
            print(i)


# Create Stack class object
stack_object = Stack()

# Pop on empty stack
stack_object.pop()

# Push elements
stack_object.push(10)
stack_object.push(20)
stack_object.push(30)
stack_object.push(50)
stack_object.push(40)
stack_object.push(70)

# Peek
print("Top element:", stack_object.peek())

# Pop on non-empty stack
print("Popped element:", stack_object.pop())

# Print stack
print("Stack elements:")
stack_object.print_stack()
```
