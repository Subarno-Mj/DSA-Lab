General Tree - A node can have any number of children

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.children = []

    def add_child(self, child):
        self.children.append(child)
# class end

root = Node("A")
b = Node("B")
c = Node("C")
d = Node("D")

root.add_child(b)
root.add_child(c)
root.add_child(d)

print("Root is -",root.data)
print([child.data for child in root.children])
```

Binary Tree - A node can have at most two children.. left and right

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None


root = Node(10)
root.left = Node(20)
root.right = Node(30)
```

Binary Tree Using List/Array - For a complete binary tree, we can use a Python list, This representation is commonly used for heaps.

```python
from collections import deque


# Node class
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None


# Create Binary Tree
root = Node(1)

root.left = Node(2)
root.right = Node(3)

root.left.left = Node(4)
root.left.right = Node(5)

root.right.left = Node(6)
root.right.right = Node(7)


# 1. Inorder Traversal
# LEFT -> ROOT -> RIGHT
def inorder(root):
    if root:
        inorder(root.left)
        print(root.data, end=" ")
        inorder(root.right)


# 2. Preorder Traversal
# ROOT -> LEFT -> RIGHT
def preorder(root):
    if root:
        print(root.data, end=" ")
        preorder(root.left)
        preorder(root.right)


# 3. Postorder Traversal
# LEFT -> RIGHT -> ROOT
def postorder(root):
    if root:
        postorder(root.left)
        postorder(root.right)
        print(root.data, end=" ")


# 4. Level Order Traversal
# Level by Level
def level_order(root):
    if root is None:
        return

    queue = deque()
    queue.append(root)

    while queue:
        node = queue.popleft()

        print(node.data, end=" ")

        if node.left:
            queue.append(node.left)

        if node.right:
            queue.append(node.right)


# Print all traversals
print("Binary Tree:")
print("       1")
print("      / \\")
print("     2   3")
print("    / \\ / \\")
print("   4  5 6  7")


print("\nInorder Traversal:")
inorder(root)

print("\nPreorder Traversal:")
preorder(root)

print("\nPostorder Traversal:")
postorder(root)

print("\nLevel Order Traversal:")
level_order(root)
```
