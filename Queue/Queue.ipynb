```python
class Queue:
    print("Queue implementation---")

    def __init__(self):
        self.queue = []

    def is_empty(self):
        if len(self.queue)==0:
            return True
        else:
            False

    def Enqueue(self,data):
        self.queue.append(data)  # enqueue at last index

    def Dequeue(self):
        if self.is_empty()==True:
            print("Queue is empty...")
            return
        else:
            return self.queue.pop(0) # dequeue at first index

    def print_queue(self):
        for i in self.queue:
            print(i, end=" ")

    def peek(self):
        if self.is_empty():
            print("Queue is empty...")
            return None
        return self.queue[0]


queue_object= Queue()

# enqueue elements
queue_object.Enqueue(122)
queue_object.Enqueue(13)
queue_object.Enqueue(14)
queue_object.Enqueue(15)
queue_object.Enqueue(155)
queue_object.Enqueue(15)
queue_object.Enqueue(17)
queue_object.Enqueue(1225)

# peek the first element
print("Peek element is =",queue_object.peek())

# Print queue
print("Queue elements:")
queue_object.print_queue()

# Dequeue
print("\nDequeued element:", queue_object.Dequeue())

# Print queue
print("Queue elements:")
queue_object.print_queue()

# peek the first element
print("\nPeek element is =",queue_object.peek())
```
